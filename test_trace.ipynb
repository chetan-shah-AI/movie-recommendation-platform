```python


from langfuse import get_client

# Initialize the global client
langfuse = get_client()

# Start an observation as a context manager
with langfuse.start_as_current_observation(
    name="my-custom-span",
    as_type="span", # Options: "span", "generation", "event", "tool", "agent"
    input={"query": "Hello world"},
) as observation:
    
    # ... Run your LLM application logic or API call here ...
    result = "Hello, user!"
    
    # (Optional) Update the observation contextually
    observation.update(output={"response": result})

# The observation automatically ends when exiting the `with` block
```
